```python
import cv2
from google.colab import files
from IPython.display import Video, display
```

```python
uploaded = files.upload()
```

```python
def reverse_video(input_video, output_video):

    # Read the video
    cap = cv2.VideoCapture(input_video)

    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Store all video frames
    frames = []

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        frames.append(frame)

    cap.release()

    # Create output video
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(
        output_video, fourcc, fps, (width, height)
    )

    # Write frames in reverse order
    for frame in reversed(frames):
        out.write(frame)

    out.release()

    print("Video reversed successfully!")

# Get uploaded video filename
input_video = list(uploaded.keys())[0]

# Reverse the video
reverse_video(input_video, "reversed_video.mp4")
```
