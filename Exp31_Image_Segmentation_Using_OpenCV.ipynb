```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
image_path = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_path)

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display original image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# Set threshold value
threshold_value = 127

# Apply binary thresholding
ret, segmented = cv2.threshold(
    gray, threshold_value, 255, cv2.THRESH_BINARY
)

# Display segmented image
plt.imshow(segmented, cmap='gray')
plt.title("Image Segmentation")
plt.axis("off")
plt.show()
```

```python
ret, inverse = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY_INV
)

plt.imshow(inverse, cmap='gray')
plt.title("Inverse Threshold Segmentation")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(segmented, cmap='gray')
plt.title("Binary Segmentation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(inverse, cmap='gray')
plt.title("Inverse Segmentation")
plt.axis("off")

plt.show()
```
